# Data ingestion and exploratory analysis

This notebook inventories local WAV recordings, inspects their metadata, and plots a short waveform and spectrogram when data is available.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

import matplotlib.pyplot as plt
import pandas as pd

from src.utils.audio_io import inspect_audio, load_audio

RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
wav_files = sorted(RAW_DIR.glob('*.wav'))
print(f'Found {len(wav_files)} WAV file(s) in {RAW_DIR}')
```

```python
metadata = pd.DataFrame([inspect_audio(path) for path in wav_files])
metadata
```

```python
if wav_files:
    audio, sample_rate = load_audio(wav_files[0])
    preview_seconds = min(5.0, len(audio) / sample_rate)
    preview = audio[:int(preview_seconds * sample_rate)]
    time = pd.Series(range(len(preview))) / sample_rate

    figure, axes = plt.subplots(2, 1, figsize=(12, 7), constrained_layout=True)
    axes[0].plot(time, preview, linewidth=0.7)
    axes[0].set(title=wav_files[0].name, xlabel='Time (s)', ylabel='Amplitude')
    axes[1].specgram(preview, Fs=sample_rate, NFFT=1024, noverlap=768, cmap='magma')
    axes[1].set(xlabel='Time (s)', ylabel='Frequency (Hz)')
    plt.show()
else:
    print('Add WAV files to data/raw/ to render waveform and spectrogram plots.')
```
